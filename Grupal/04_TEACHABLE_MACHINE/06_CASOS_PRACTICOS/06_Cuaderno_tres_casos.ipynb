## Guía de predicción real (solo después de entrenar)

Cada equipo debe entrenar en Teachable Machine con sus fotos propias, exportar `keras_model.h5` y `labels.txt` y **reservar imágenes completamente nuevas** para probar el clasificador. Al probar, registrar tanto aciertos como errores. Una confianza mayor no prueba mejor exactitud.

```python
# Función de análisis disponible únicamente cuando hay un modelo real exportado.
# Este código NO fabrica resultados ni genera un archivo .h5.
def revisar_exportacion(carpeta):
    carpeta = Path(carpeta)
    modelo = carpeta / 'modelo' / 'keras_model.h5'
    etiquetas = carpeta / 'modelo' / 'labels.txt'
    if not (modelo.is_file() and etiquetas.is_file()):
        return {'estado': 'PENDIENTE', 'detalle': 'Entrenar y exportar desde Teachable Machine.'}
    clases = [x.strip() for x in etiquetas.read_text(encoding='utf-8').splitlines() if x.strip()]
    return {'estado': 'ARCHIVOS EXPORTADOS', 'clases': clases, 'modelo_bytes': modelo.stat().st_size}
for nombre, carpeta in CASOS.items():
    print(nombre, revisar_exportacion(carpeta))
```
